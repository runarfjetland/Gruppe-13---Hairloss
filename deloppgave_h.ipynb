```python
arstall = input("Skriv inn ett årstall fra og med 2014, til og med 2025: ")

somerdager = 0
hoysommerdager = 0
tropedager = 0

with open("sinnes_2014_2025_med_makstemperatur.csv", "r") as file:
    for linje in file:
        deler = linje.strip().split(";")
        dato = deler[2].split(".")
        if "." in deler[2]:
            if dato[2] == arstall:
                if deler[3] >= "20" and deler[3] < "25":
                    somerdager += 1
                elif deler[3] >= "25" and deler[3] < "30":
                    hoysommerdager += 1
                elif deler[3] >= "30":
                    tropedager += 1
                else:
                    pass
    print(f"Antall sommerdager i {arstall}: {somerdager}")
    print(f"Antall høysommerdager i {arstall}: {hoysommerdager}")
    print(f"Antall tropedager i {arstall}: {tropedager}")
```
