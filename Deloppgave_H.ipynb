```python
import matplotlib.pyplot as plt

aar = []
sommer = []
hoysommer = []
trope = []

for arstall in range(1954, 2026):
    sommerdager = 0
    hoysommerdager = 0
    tropedager = 0

    with open("oslo_blindern_1954_2025_med_makstemperatur.csv", "r") as file:
        for linje in file:
            deler = linje.strip().split(";")

            if "." in deler[2]:
                dato = deler[2].split(".")

                if dato[2] == str(arstall):
                    if deler[3] != "-":
                        temp = float(deler[3].replace(",", "."))

                        if temp > 20:
                            sommerdager += 1
                        if temp > 25:
                            hoysommerdager += 1
                        if temp > 30:
                            tropedager += 1

    aar.append(arstall)
    sommer.append(sommerdager)
    hoysommer.append(hoysommerdager)
    trope.append(tropedager)

plt.plot(aar, sommer, label="Sommerdager")
plt.plot(aar, hoysommer, label="Høysommerdager")
plt.plot(aar, trope, label="Tropedager")

plt.xlabel("År")
plt.ylabel("Antall dager")
plt.title("Temperatur Oslo Blindern")
plt.legend()
plt.grid()
plt.show()
```
